```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="moonshotai/kimi-k2-instruct-0905",
    temperature=0,
    max_tokens=None,
    timeout=None,
    max_retries=2,
    # other params...
)
```

```python
from langchain_community.document_loaders import WebBaseLoader

loader = WebBaseLoader("https://usijobs.deloitte.com/en_US/careersUSI/JobDetail/USI-EH26-Consulting-AI-Data-Consultant-Python-Data-Engineer/300054")
page_data = loader.load().pop().page_content
print(page_data)
```

```python
from langchain_core.prompts import PromptTemplate

prompt_extract = PromptTemplate.from_template(
        """
        ### SCRAPED TEXT FROM WEBSITE:
        {page_data}
        ### INSTRUCTION:
        The scraped text is from the career's page of a website.
        Your job is to extract the job postings and return them in JSON format containing the
        following keys: `role`, `experience`, `skills` and `description`.
        Only return the valid JSON.
        ### VALID JSON (NO PREAMBLE):
        """
)

chain_extract = prompt_extract | llm
res = chain_extract.invoke(input={'page_data':page_data})
type(res.content)
```

```python
from langchain_core.output_parsers import JsonOutputParser

json_parser = JsonOutputParser()
json_res = json_parser.parse(res.content)
json_res
```

```python
import pandas as pd

df = pd.read_csv("/content/student portfolios.csv")
df
```

```python
import uuid
import chromadb

client = chromadb.PersistentClient('vectorstore')
collection = client.get_or_create_collection(name="portfolio")

if not collection.count():
    for _, row in df.iterrows():
        collection.add(documents=row["Techstack"],
                       metadatas={"links": row["Links"]},
                       ids=[str(uuid.uuid4())])
```

```python
link_list = collection.query(query_texts=job['skills'], n_results=2).get('metadatas', [])
link_list
```

```python
prompt_email = PromptTemplate.from_template(
        """
        ### JOB DESCRIPTION:
        {job_description}

        ### INSTRUCTION:
        You are Sam Smith, Business Development Executive at Bridge INC, a Consulting and Talent Solutions firm that helps companies hire top professionals across all domains.

        Write a cold email to a potential client about their job role at company_name, introducing Bridge INC’s ability to provide qualified talent for the position.
        Include relevant portfolio links from {link_list}, and end with a call-to-action to schedule a brief discussion.
        Keep the tone professional and persuasive.

        Sam Smith
        Business Development Executive | Bridge INC
        sam.smith@bridgeinc.com | +916465550198
        www.bridge-inc.com
        """
        )

chain_email = prompt_email | llm
res = chain_email.invoke({"job_description": str(job), "link_list": link_list})
print(res.content)
```
